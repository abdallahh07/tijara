```python
import pandas as pd 
import numpy as np 
```

```python
data = pd.read_csv("../data/output/user_events.csv")
```

```python
data.head()
```

```python
data["eventType"].value_counts()
```

```python
event_weights={
    "VIEW":1,
    "ADD_TO_CART":4,
    "SEARCH":2,
    "WISHLIST_ADD":3,
    "REMOVE_FROM_CART":-1,
    "PURCHASE":5}
```

```python
data["weight"] = data["eventType"].map(event_weights)
```

```python
data
```

```python
interaction_score = data.groupby(["userId","productId"])["weight"].sum().reset_index()
```

```python
interaction_score["weight"].value_counts()
```

```python
unique_users = interaction_score["userId"].unique()
unique_products = interaction_score["productId"].unique()
```

```python
len(unique_users),len(unique_products)
```

```python
user_idx = {user_id: i for i,user_id in enumerate(unique_users)}
product_idx = {product_id: i for i,product_id in enumerate(unique_products)}
```

```python
interaction_score["user_idx"] = interaction_score["userId"].map(user_idx)
interaction_score["product_idx"] = interaction_score["productId"].map(product_idx)
```

```python
interaction_score
```

## the model

```python
from scipy.sparse import csr_matrix
```

```python
interaction_matrix= csr_matrix(
    (interaction_score["weight"], (interaction_score["user_idx"], interaction_score["product_idx"])),
    shape=(len(unique_users), len(unique_products)))
```

## Model Choice: Matrix Factorization (ALS) via `implicit`

This recommender uses Alternating Least Squares (ALS) matrix factorization, 
implemented via the `implicit` library — a classical collaborative filtering 
technique, not deep learning. ALS was chosen because:
- It's well-suited to implicit feedback data (views, cart adds, purchases) 
  rather than explicit ratings
- It's fast to train and interpret, appropriate for an MVP recommendation system
- It's a widely-used, production-proven approach (used historically by 
  companies like Spotify for exactly this kind of implicit-feedback problem)

A future iteration could explore neural collaborative filtering, but ALS is 
a strong, simpler baseline for this stage of the project.

```python
from implicit.als import AlternatingLeastSquares
model = AlternatingLeastSquares(factors=50)
model.fit(interaction_matrix)
```

```python
recommended = model.recommend(0, interaction_matrix[0])
recommended
```

```python
index_to_product = {}
for product_id, index in product_idx.items():
    index_to_product[index] = product_id
```

```python
index_to_product[66]
```

```python
recommended_indices = recommended[0]
recommended_scores = recommended[1]

recommended_product_ids = []
for i in recommended_indices:
    product_id = index_to_product[i]
    recommended_product_ids.append(product_id)

for pid, score in zip(recommended_product_ids, recommended_scores):
    print(pid, score)
```
