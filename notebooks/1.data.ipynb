# 1. Generate Categories

## Category Table

Matches the `Category` table in the production Prisma schema. No foreign key 
dependencies — generated first.

| Column     | Type            | Notes                                      |
|------------|-----------------|---------------------------------------------|
| `id`       | String (PK)     | UUID, generated via `uuid.uuid4()`          |
| `name`     | String          | Category display name                       |
| `slug`     | String          | Unique, URL-friendly (lowercase, hyphenated)|
| `parentId` | String? (FK)    | Nullable — supports nested categories, set to `None` for this initial batch (flat categories only, no sub-categories yet) |

**Generated rows:** 8 top-level categories — Electronics, Clothing, Home and 
Garden, Beauty and Personal Care, Sports and Outdoors, Books, Toys and Games, 
Groceries.

**Output:** saved to `data/output/categories.csv`m

```python
category_names = [
    "Electronics", "Clothing", "Home and Garden",
    "Beauty and Personal Care", "Sports and Outdoors",
    "Books", "Toys and Games", "Groceries"
]

categories = []
for name in category_names:
    category = {
        "id": str(uuid.uuid4()),        # ← this is what you just tested
        "name": name,
        "slug": name.lower().replace(" ", "-"),
        "parentId": None
    }
    categories.append(category)

categories
```

```python
categories_df = pd.DataFrame(categories)
categories_df
```

```python
categories_df.to_csv("../data/output/categories.csv", index=False)
```

# 2. Generate Brands

Matches the `Brand` table in the production Prisma schema. No foreign key 
dependencies — generated independently of Category (they only connect later, 
through `Product`, which references both `categoryId` and `brandId`).

| Column    | Type         | Notes                                  |
|-----------|--------------|------------------------------------------|
| `id`      | String (PK) | UUID, generated via `uuid.uuid4()`      |
| `name`    | String       | Brand display name                       |
| `slug`    | String       | Unique, URL-friendly (lowercase, hyphenated) |
| `logoUrl` | String?      | Nullable — set to `None` for now, no real logos yet |

**Output:** saved to `data/output/brands.csv`

```python
brand_names = [
    "Nova", "Zenith", "Urban Edge", "Pure Living", "Apex Gear",
    "Bright Home", "Trendify", "Everyday Essentials", "Peak Performance",
    "Lumin", "Craft & Co", "Vitality", "Solstice", "Momentum"
]
```

```python
brands = []
for name in brand_names:
    brand = {
        "id":str(uuid.uuid4()),
        "name":name,
        "slug": name.lower().replace(" ", "-").replace("&", "").replace("--", "-"),
        "logoUrl": None}
    brands.append(brand)
```

```python
brands
```

```python
brands_df = pd.DataFrame(brands)
brands_df
```

```python
brands_df.to_csv("../data/output/brands.csv", index=False)
```

# 3. Generate Products

```python
num_products = 150
products = [] 
for i in range(num_products):
    category = categories[np.random.randint(0,len(categories))]
    brand = brands[np.random.randint(0,len(brands))]

    product = {
        "id": str(uuid.uuid4()),
        "name":f"{category['name']} Product {i}",
        "slug": f"{category['name']} Product {i}".lower().replace(" ", "-"),
        "description": f"A great {category['name']} product from {brand['name']}.",
        "categoryId":category["id"],
        "brandId":brand["id"],
        "basePrice": np.random.uniform(1000, 10000),
        "createdAt": "2026-01-01",
        "updatedAt": "2026-01-01"}
    products.append(product)
```

# 5.UserEvent

```python
event_types = ["VIEW", "ADD_TO_CART", "SEARCH", "WISHLIST_ADD", "REMOVE_FROM_CART", "PURCHASE"]
weights = [0.60, 0.15, 0.10, 0.05, 0.05, 0.05]

[np.random.choice(event_types, p=weights) for _ in range(20)]
```

```python
import datetime
```

```python
num_events = 37000
user_events = []

for i in range(num_events):
    user = users[np.random.randint(0, len(users))]
    product = products[np.random.randint(0, len(products))]
    event_type = np.random.choice(event_types, p=weights)
    random_days_ago = int(np.random.randint(0, 90))
    event_date = datetime.date.today() - datetime.timedelta(days=random_days_ago)
    
    event = {
        "id": str(uuid.uuid4()),
        "userId": user["id"],
        "sessionId": str(uuid.uuid4()),
        "productId": product["id"],
        "eventType": event_type,
        "metadata": None,
        "createdAt": event_date
    }
    user_events.append(event)
```

```python
user_event = pd.DataFrame(user_events)
```

```python
user_event
```

```python
user_event.to_csv("../data/output/user_events.csv", index=False)
```

```python
purchase_events = [e for e in user_events if e["eventType"] == "PURCHASE"]
len(purchase_events)
```

```python
purchase_events = [e for e in user_events if e["eventType"] == "PURCHASE"]

orders = []
order_items = []

for pe in purchase_events:
    variant = variants[np.random.randint(0, len(variants))]
    quantity = int(np.random.randint(1, 4))
    order_id = str(uuid.uuid4())

    orders.append({
        "id": order_id,
        "userId": pe["userId"],
        "status": "DELIVERED",
        "createdAt": pe["createdAt"]
    })

    order_items.append({
        "id": str(uuid.uuid4()),
        "orderId": order_id,
        "variantId": variant["id"],
        "quantity": quantity
    })

orders_df = pd.DataFrame(orders)
order_items_df = pd.DataFrame(order_items)
```

```python
orders_df.to_csv("../data/output/orders.csv", index=False)
order_items_df.to_csv("../data/output/order_items.csv", index=False)
```

```python
# Price lookup (needed before DailyProductMetrics)
product_price_lookup = {}
for p in products:
    product_price_lookup[p["id"]] = p["basePrice"]

# Build DailyProductMetrics from UserEvent view data
user_event_df = pd.DataFrame(user_events)
user_event_df['date'] = pd.to_datetime(user_event_df['createdAt'])

view_counts = (
    user_event_df[user_event_df['eventType'] == 'VIEW']
    .groupby(['productId', 'date'])
    .size()
    .reset_index(name='viewCount')
)

daily_metrics = []
for _, row in view_counts.iterrows():
    daily_metrics.append({
        "id": str(uuid.uuid4()),
        "productId": row["productId"],
        "date": row["date"].date(),
        "unitsSold": int(np.random.randint(0, 10)),
        "revenue": 0,
        "avgPrice": product_price_lookup.get(row["productId"], 0),
        "viewCount": row["viewCount"],
        "stockEndOfDay": int(np.random.randint(0, 200))
    })

daily_metrics_df = pd.DataFrame(daily_metrics)
print(len(daily_metrics_df))
daily_metrics_df.head()
```
