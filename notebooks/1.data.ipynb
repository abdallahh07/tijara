# 1. Generate Categories

## Category Table

Matches the `Category` table in the production Prisma schema. No foreign key 
dependencies — generated first.

| Column     | Type            | Notes                                      |
|------------|-----------------|---------------------------------------------|
| `id`       | String (PK)     | UUID, generated via `uuid.uuid4()`          |
| `name`     | String          | Category display name                       |
| `slug`     | String          | Unique, URL-friendly (lowercase, hyphenated)|
| `parentId` | String? (FK)    | Nullable — supports nested categories, set to `None` for this initial batch (flat categories only, no sub-categories yet) |

**Generated rows:** 8 top-level categories — Electronics, Clothing, Home and 
Garden, Beauty and Personal Care, Sports and Outdoors, Books, Toys and Games, 
Groceries.

**Output:** saved to `data/output/categories.csv`m

```python
category_names = [
    "Electronics", "Clothing", "Home and Garden",
    "Beauty and Personal Care", "Sports and Outdoors",
    "Books", "Toys and Games", "Groceries"
]

categories = []
for name in category_names:
    category = {
        "id": str(uuid.uuid4()),        # ← this is what you just tested
        "name": name,
        "slug": name.lower().replace(" ", "-"),
        "parentId": None
    }
    categories.append(category)

categories
```

```python
categories_df = pd.DataFrame(categories)
categories_df
```

# 2. Generate Brands

Matches the `Brand` table in the production Prisma schema. No foreign key 
dependencies — generated independently of Category (they only connect later, 
through `Product`, which references both `categoryId` and `brandId`).

| Column    | Type         | Notes                                  |
|-----------|--------------|------------------------------------------|
| `id`      | String (PK) | UUID, generated via `uuid.uuid4()`      |
| `name`    | String       | Brand display name                       |
| `slug`    | String       | Unique, URL-friendly (lowercase, hyphenated) |
| `logoUrl` | String?      | Nullable — set to `None` for now, no real logos yet |

**Output:** saved to `data/output/brands.csv`

```python
brand_names = [
    "Nova", "Zenith", "Urban Edge", "Pure Living", "Apex Gear",
    "Bright Home", "Trendify", "Everyday Essentials", "Peak Performance",
    "Lumin", "Craft & Co", "Vitality", "Solstice", "Momentum"
]
```

```python
brands = []
for name in brand_names:
    brand = {
        "id":str(uuid.uuid4()),
        "name":name,
        "slug": name.lower().replace(" ", "-").replace("&", "").replace("--", "-"),
        "logoUrl": None}
    brands.append(brand)
```
